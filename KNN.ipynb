KNN is a simple supervised machine learning algorithm used for both classification, regression, Pattern Recognition and Recommendation Systems
It predicts the output of a new data point based on the majority or average of its nearest neighbors.
Example: 1. movie recommendations
         2. Image recognition

KNN (K-Nearest Neighbors) working
step1: Choose the value of K (number of neighbors).
step2: Find the K nearest neighbors.
step3: Check the majority class among the nearest neighbors.
step4: Assign the class with the highest votes as the prediction.
